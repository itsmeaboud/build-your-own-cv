```python
# ===============================
# 1. Import libraries
# ===============================
import numpy as np
import cv2
import matplotlib.pyplot as plt

import sys
sys.path.append("..")
from core.transformation import warpAffine , getRotationMatrix2D
from core.utils.visualization import show_two_images, show_image
from core.utils.image_io import load_image
```

```python
# ===============================
# 2. Load the image
# ===============================
img_path = "../data/images/cameraman.jpeg"
img = load_image(img_path)

show_image(img)
img.shape
```

```python
# ===============================
# 3. Generating Rotation Matrix
# ===============================
M_custom = getRotationMatrix2D((0,0), 30, 1)
M_cv = cv2.getRotationMatrix2D((img.shape[0] / 2, img.shape[1] / 2),
                                30,
                                1)

print("Shape of M_custom:", M_custom.shape)
print(M_custom)
print("==========================================")
print("Shape of M_cv:", M_cv.shape)
print(M_cv)
```

```python
# ===============================
# 4. Applying rotation
# ===============================
rotated_img_custom = warpAffine(img, M_custom)
rotated_img_cv = cv2.warpAffine(img, M_cv, img.shape[:2])
```

```python
show_two_images(rotated_img_custom, 
                rotated_img_cv, 
                "Custom", 
                "CV"
               )
```
